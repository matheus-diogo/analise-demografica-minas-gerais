# Notebook para Geração dos Gráficos utilizados no Relatório

## Importação dos Pacotes Necessários

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Razão de Sexo em 2010 e 2022

```python
# Importação dos dados
rs_2010_2022 = pd.read_excel('DadosGráficos.xlsx', 'Razão_Sexo_Ano')

# Geração do gráfico de barras
plt.figure(figsize=(10,10))
sns.barplot(
    data = rs_2010_2022,
    x = 'Ano',
    y = 'RS',
    color = '#ffdf87',
    zorder = 2
)
plt.grid(alpha = 0.4, axis = 'y')
plt.ylabel('Razão de Sexo')
plt.show()
```

## Pirâmide Etária em 2010

```python
# Importação dos dados
pr_2010 = pd.read_excel('DadosGráficos.xlsx', 'População_Residente_2010_Sexo')

# Transformação dos dados
pr_2010['Homens'] = - pr_2010['Homens']

# Geração da pirâmide etária
plt.figure(figsize=(13,13))
ax1 = sns.barplot(
    data = pr_2010,
    x = 'Homens',
    y = 'Idade',
    order = pr_2010['Idade'][::-1],
    color = '#aae7fa',
    zorder=2
)
ax2 = sns.barplot(
    data = pr_2010,
    x = 'Mulheres',
    y = 'Idade',
    order = pr_2010['Idade'][::-1],
    color = '#faaae5',
    zorder=2
)
plt.xlabel('Homens/Mulheres')
plt.ylabel('Faixa Etária')
plt.grid(alpha=0.4, axis='x')
plt.xticks(
    ticks=[-800000, -600000, -400000, -200000, 0, 200000, 400000, 600000, 800000],
    labels=['800k', '600k', '400k', '200k', '0', '200k', '400k', '600k', '800k']
)
plt.show()
```

## Evolução dos Óbitos e dos Nascimentos entre 2010 e 2022

```python
# Importação dos dados
obitos_nascimentos = pd.read_excel('DadosGráficos.xlsx', 'Óbitos_Nascimentos_Ano')

# Transformação dos dados
obitos_nascimentos = obitos_nascimentos.melt(
    id_vars = 'Ano',
    value_vars = ['Óbitos', 'Nascidos Vivos', 'CN'],
    var_name = 'Classificação',
    value_name = 'Quantitativo'
)

# Geração do gráfico de linhas
plt.figure(figsize=(10,5))
sns.lineplot(
    data = obitos_nascimentos,
    x = 'Ano',
    y = 'Quantitativo',
    hue = 'Classificação',
    marker = 'o',
    palette = ['#cf726b', '#5fad62', '#738ec9'],
    zorder = 5
)
plt.grid(alpha = 0.4)
plt.axvline(x=2016, color='black', linestyle='--')
plt.axvline(x=2021, color='black', linestyle='--')
plt.show()
```

## Distribuição das Probabilidades de Óbito por Faixa Etária e Sexo

```python
# Importação dos dados
probabilidades_obito = pd.read_excel('DadosGráficos.xlsx', 'Probabilidades_Óbito_Sexo_2022')

# Transformação dos dados
probabilidades_obito = probabilidades_obito[probabilidades_obito['Faixa'] != '[80,+)']
probabilidades_obito = probabilidades_obito.melt(
    id_vars = 'Faixa',
    value_vars = ['Homens','Mulheres'],
    var_name = 'Sexo',
    value_name = 'Probabilidade'
)

# Geração do gráfico de barras horizontais
plt.figure(figsize=(10,10))
sns.barplot(
    data = probabilidades_obito,
    x = 'Probabilidade',
    y = 'Faixa',
    hue = 'Sexo',
    palette = ['#aae7fa', '#faaae5'],
    order = probabilidades_obito['Faixa'][::-1],
    zorder = 2
)
plt.ylabel('Faixa Etária')
plt.grid(alpha = 0.4, axis = 'x')
plt.show()
```

## Nascivos Vivos por Faixa Etária da Mãe

```python
# Importação dos dados
nascidos_vivos_faixa_mae = pd.read_excel('DadosGráficos.xlsx', 'Nascidos_Vivos_Idades_Mães_Ano')

# Transformação dos dados
nascidos_vivos_faixa_mae = nascidos_vivos_faixa_mae.melt(
    id_vars = 'Faixa',
    value_vars = [2010, 2022], 
    var_name = 'Ano', 
    value_name = 'Nascimentos'
)

# Geração do gráfico de barras
plt.figure(figsize=(10,7))
sns.barplot(
    data = nascidos_vivos_faixa_mae,
    x = 'Faixa',
    y = 'Nascimentos',
    hue = 'Ano',
    palette = ['#b3d7ae', '#44935b'],
    zorder = 2
)
plt.xlabel('Faixa Etária')
plt.grid(alpha = 0.4, axis = 'y')
plt.show()
```

## Maiores Migrações Absolutas para Minas Gerais em 2010 e 2022

```python
# Importação dos dados
migracao_absoluta_para_mg = pd.read_excel('DadosGráficos.xlsx', 'Migração_Absoluta_Para_MG_Ano')

# Transformação dos dados
migracao_absoluta_para_mg = migracao_absoluta_para_mg.head(5)
migracao_absoluta_para_mg = migracao_absoluta_para_mg.melt(
    id_vars = 'Lugar de Nascimento',
    value_vars = ['2010','2022'],
    var_name = 'Ano',
    value_name = 'Migração Absoluta'
)

# Geração do gráfico de barras
plt.figure(figsize=(10,6))
sns.barplot(data = migracao_absoluta_para_mg, x = 'Lugar de Nascimento', y = 'Migração Absoluta', hue = 'Ano', palette = 'Blues', zorder = 2)
plt.ylabel('Migração Absoluta')
plt.grid(alpha = 0.4, axis = 'y')
plt.show()
```

## Migração Interna em Minas Gerais em 2010 e 2022 

```python
# Importação dos dados
migracao_interna_mg_2010_2022 = pd.read_excel('DadosGráficos.xlsx', 'Migração_Interna_MG_Ano')

# Transformação dos dados
migracao_interna_mg_2010_2022 = migracao_interna_mg_2010_2022.melt(
    id_vars = 'Ano',
    value_vars=['Imigrantes','Emigrantes'],
    var_name='Classificação',
    value_name='Quantitativo'
)

# Geração do gráfico de barras
plt.figure(figsize=(10,6))
sns.barplot(
    data = migracao_interna_mg_2010_2022, 
    x = 'Ano', 
    y = 'Quantitativo', 
    hue = 'Classificação', 
    palette = ['#53e653','#e65553'],
    zorder = 2
).ticklabel_format(style='plain', axis='y')
plt.grid(alpha=0.4, axis='y')
plt.show()
```
